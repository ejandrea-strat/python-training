# 1. Comprehensions

Builds a new collection (list, set, dict, or generator) from an esiting iterale in one expression.

## Syntax

[expression for item in iterable if condition]

```python
scores = [72, 95, 58, 88, 64]

# Using if on the right side kinda acts like filter()
passed = [s for s in scores if s >= 65]

# Putting the logic on the left side kinda acts like map()
curved = [s + 5 for s in scores]          

# I thought this would add 5 first then filter, but it's the other way around
adjusted = [s + 5 for s in scores if s >= 65]

print(f'Passed: {passed}')
print(f'Curved: {curved}')
print(f'Adjusted: {adjusted}')


# So I don't get dizzy with the syntax, I read it starting from the for part first then left side next

printed_scores = [print(s) for s in scores]
print(printed_scores) # printing appears to return None in a comprehension
# Upon further research, this is because the print() function itself returns None, and comprehensions just store whatever the expression evaluates to.
# I researched more Python functions that return None, and most of them turned out to be functions that are called for a side effect (an action or an in-place change,
# like printing, appending or sorting).
```

## Loop

```python
# Turns out the order I read it in is also how I'd rewrite it as a loop.
# To me right now, a comprehension is a compressed loop that builds a new collection from any iterable.

passed = []

for s in scores:
    if s >= 65:
        passed.append(s)

print(f'Passed: {passed}')
```

## Four kinds of Comprehension

```python
words = 'HELLO'

print(f'Square Brackets builds a list: {[w.lower() for w in words]}')

# drops duplicates, since there's two l's, only one remains 
# (the order came out shuffled too because they're unordered collections)
print(f'Curly brackets build a set: { {w.lower() for w in words} }')

# keys can't repeat, the second 'L' overwrites the first
print(f'Curly brackets with key:value build a dict: { {w: len(w) for w in words} }') 

print('Parenthesis build a generator')
# a generator is like a paused function that it gives one value each time it's asked,
# and once it runs out, it's used up.
gen = (n * 10 for n in [1, 2, 3])
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen)) # throws an error because there's no item to iterate to
```

## Filter vs Transform

```python
# The if logic earlier, basically it acted as a filter
[s for s in scores if s >= 65]

# Returned each values but this time replaced the items with values of pass or fail based on the logic
print(["pass" if s >= 65 else "fail" for s in scores])

# My earlier observation stands:
# if the logic is on the right side, it's a filter; if it's on the left side, it's a map/transform.
```

## Nested Comprehensions

```python
grid = [[1, 2], [3, 4], [5, 6]]
flat = [n for row in grid for n in row]

# A bit hard to read at first. With two fors, the first one is the outer loop and the next one runs inside it:
# row 1 item 1, row 1 item 2, row 2 item 1, row 2 item 2, and so on.

print(f'Flattened values of the grid: {flat}')

cube = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
flat = [n for layer in cube for row in layer for n in row]

# Since I got used to the first example, this one is easier to read, but the syntax still gets confusing sometimes.
# Layer -> row -> numbercan

print(f'Flattened values of the cube: {flat}')


table = [[r * c for c in range(1, 4)] for r in range(1, 4)]
# The outer range takes a value first, for instance 1.
# Then the inner range runs through all its values (1, 2, 3), multiplying each by the outer value and that builds one row
# Once the inner range runs out, the outer range takes its next value, 2, and the inner range starts over from 1
print(table)
```

## Iterables

Anything a for loop can walk through. They are lists, tuples, strings, dictionaries, sets, and ranges.

```python
prices = {"pen": 12.0, "notebook": 45.5}
sale = {name: round(price * 0.9, 2) for name, price in prices.items()}

print(sale)

for i, name in enumerate(["a", "b"], start=1):
    print(i, name)
```

## map() and filter()

```python
print(f'Map: {list(map(str.upper, ["a", "b"]))}') # transforms the values
print(f'Filter: {list(filter(str.isdigit, ["12", "x", "7"]))}') # filters the values

# Just found out that map, filter, and generators have a lot in common:
# they're all single-use, lazy iterators (nothing is computed until something asks,
# and they hand out values one at a time).
# The values show up here only because I wrapped each one in list(), which asks for all of them at once.
```

# Exceptions

How python reports something went wrong while running and running and what to do about it.

- `Syntax error`: Python can't even read the code, like a missing colon or an unclosed bracket. Nothing runs until you fix it, and you can't catch it.
- `Exception`: the code is valid, but something goes wrong while it runs. These can be caught and handled.

```python
int("12a")
# raises a ValueError: invalid literal for int() with base 10: '12a'

# The last line is the exception type and message. The lines above it show where it happened, 
# ending with the line that failed ("most recent call last").
```

## try / except / else / finally

```python
stock = {"pen": 40, "eraser": 0}

def units_per_box(item, boxes):
    try:
        per_box = stock[item] / boxes      # the only line that can fail
    except KeyError:
        print(f"unknown item: {item}")     # runs only if KeyError happened
    except ZeroDivisionError:
        print("boxes can't be zero")
    else:
        print(f"{per_box:.1f} per box")    # runs only if nothing failed
    finally:
        print(f"checked {item}")           # always runs



units_per_box("pen", 2) # Successful case
units_per_box("pen", 0)  # Zero Division case
units_per_box("pencil", 2)  # KeyError case
units_per_box("eraser", 5)  # Zero Items in stock case

'''
Keep the try block small: only the line that can fail. Then you know exactly what you're catching.
else holds what should happen only on success. Putting that code inside try would also catch its errors by accident.
finally is for cleanup that must happen no matter what.
'''
```

# 1.4 Text Files

## open() and modes

- `r` (read, default): reads the file from the start, if it doesn't exist, it throws a FileNotFoundError
- `w` (write): empties it first
- `a` (append): adds to the end of the file
- `x` (create only): if the file exists, it throws a FileExistsError

if the file doesn't exist, w, a, and x all creates it.

add `b` for binary: `rb`, `wb` 

```python
!touch notes.txt
!echo "hello" > notes.txt

f = open("notes.txt", "r", encoding="utf-8")
print(f.read())
```

## text vs binary

- `text`: you get string. Python decodes the bytes on disk using an encoding, and turns Windows line endings into \n
- `binary mode (rb)`: you get raw bytes, untouched. Use it for images, compressed files and formats like Avro.

## encoding

Text files are bytes on disk, and the encoding is the mapping between bytes and characters. UTF-8 is the standard. 

- Always pass `encoding="utf-8"` Without it, Python uses your system's default, which differs between Windows, Mac and Linux. The same code can then read a file fine on one machine and garble it on another.
- `UnicodeDecodeError` means the file isn't in the encoding you told Python.

## the with statement

files are operating-system resources, and writes can sit in a bubffer until the file is closed.
`with` guarantees the close, like a built-in `try/finally`

```python
with open("notes.txt", encoding="utf-8") as f:
    text = f.read()
# the file is closed here, even if an error happened inside the block
```

## Reading: all at once vs line by line

- Line by line keeps memory flat however big the file is, and lets you handle each line as it comes.
- Every line keeps its newline. line.strip() removes it, along with surrounding spaces. A blank line becomes "" after stripping.
- A file handle only reads forward. Read it a second time and you get "". Reopen the file instead.

```python
with open("server.log", encoding="utf-8") as f:
    whole = f.read()          # one big string

with open("server.log", encoding="utf-8") as f:
    lines = f.readlines()     # a list of lines, each still ending in "\n"

with open("server.log", encoding="utf-8") as f:
    for line in f:            # one line at a time
        ...
```

## writing

For CSV files, use the standard csv module instead of joining strings with commas. It handles values that contain commas or quotes. Open CSV files with newline="", as its docs say.

```python
with open("summary.txt", "w", encoding="utf-8") as f:
    f.write("first line\n")          # write() never adds the newline for you
    print("second line", file=f)     # print() does
```

## paths, and paths inside a container

- A relative path like data/notes.txt starts from the current working directory. In Jupyter that's usually the notebook's folder, and Path.cwd() shows it.
- pathlib.Path builds paths that work on any OS: Path("data") / "notes.txt", plus .exists(), .read_text() and .write_text().
- The container has its own filesystem. A bind mount, one of the volumes: lines in the compose file, makes a laptop folder appear inside the container: same files, two paths. For example, ./data:/home/jovyan/data means de-training/data on your laptop is /home/jovyan/data in the container.
- Write to a mounted path and the file appears on your laptop. Write anywhere else and it only lives in the container, and it's gone when the container is recreated.
